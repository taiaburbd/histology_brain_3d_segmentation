```python
import nibabel as nib

def get_nifti_bit_depth(nifti_image_path):
    # Load the NIfTI image
    img = nib.load(nifti_image_path)
    print(img.shape)
    # Get the data type of the image
    data_type = img.get_data_dtype()
    
    # Determine the bit depth based on the data type
    if data_type.kind in ['u', 'i','f']:  # unsigned integers or signed integers
        return data_type.itemsize * 8
    else:
        raise ValueError("Unsupported data type for bit depth calculation: {}".format(data_type))

# Example usage
nifti_image_path = '/user1/ngmm/tr855969/Desktop/Taiabur/ngmm-nnunet/dataset/nnUNet_raw_data/Dataset001_Wdr47Kusss/imagesTr/Wdr47Kusss_NG_2603_0000.nii.gz'
bit_depth = get_nifti_bit_depth(nifti_image_path)
print(f"The NIfTI image is {bit_depth}-bit.")
```

```python
import nrrd
import nibabel as nib
import numpy as np
from scipy.ndimage import zoom
import matplotlib as plt

def downsample_and_save_nrrd_to_nifti(nrrd_path, nifti_path):
    # Read the .nrrd file
    data, header = nrrd.read(nrrd_path)
    x_size, y_size, z_size = data.shape
    print(x_size, y_size, z_size)
    # Downsample the data by a factor of 2^5 = 32
    # This factor is an example, adjust according to your needs
    # Note: This simplistic downsampling does not apply any smoothing or filtering
    # For better quality downsampling, consider using more sophisticated methods
    downsampled_data = zoom(data, zoom=(0.5**3, 0.5**3, 0.5**3), order=1) # order=1 (linear interpolation)
    
    # Convert the downsampled data to 32-bit float
    downsampled_data_float32 = downsampled_data.astype(np.float32)
    
    # Create a NIfTI image; assume voxel size is 1.0, 1.0, 1.0
    affine = np.diag([1.0, 1.0, 1.0, 1])
    nifti_img = nib.Nifti1Image(downsampled_data_float32, affine)
    
    # Save the NIfTI image
    nib.save(nifti_img, nifti_path)
```

```python

# Example usage
nrrd_path = '/work/shared/ngmm/3Dimage/NG2567_8bits.nrrd'
nifti_path = '/user1/ngmm/tr855969/Desktop/Taiabur/ngmm-nnunet/dataset/nnUNet_test_data/raw/NG2567_L2.nii.gz'
downsample_and_save_nrrd_to_nifti(nrrd_path, nifti_path)

bit_depth = get_nifti_bit_depth(nifti_path)
print(f"The NIfTI image is {bit_depth}-bit.")
```

```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
import os

def save_and_plot_roi_masks(roi_image_path, output_dir, slice_index):
    # Ensure the output directory exists
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # Load the ROI mask 3D image
    roi_img = nib.load(roi_image_path)
    roi_data = roi_img.get_fdata()
    
    # Find unique region identifiers, ignoring the background (0)
    regions = np.unique(roi_data)[1:]  # Skip the first element as it's 0 (background)
    
    # Placeholder for combined 2D view of all regions
    combined_slice = np.zeros_like(roi_data[:,:,slice_index])
    
    # Process each region
    for region in regions:
        # Create a mask for the current region
        region_mask = np.where(roi_data == region, 1, 0)
        
        # Save the mask as a new NIfTI image
        region_img = nib.Nifti1Image(region_mask, roi_img.affine, roi_img.header)
        nib.save(region_img, os.path.join(output_dir, f'region_{int(region)}.nii.gz'))
        
        # Update the combined slice for 2D visualization
        combined_slice += (region_mask[:,:,slice_index] * region)
    
    # Plotting the combined 2D view of all regions
    plt.imshow(combined_slice, cmap='tab20', interpolation='nearest')
    plt.title('Combined 2D view of all regions')
    plt.colorbar()
    plt.axis('off')
    plt.show()

# Example usage
roi_image_path = '/work/shared/ngmm/scripts/Taiabur/NG2564_SegmentsSeeds.nii'
output_dir = '/user1/ngmm/tr855969/Desktop/Taiabur/ngmm-nnunet/dataset/individual_regions/NG2564'
slice_index = 201  # Change this based on which slice you want to visualize
save_and_plot_roi_masks(roi_image_path, output_dir, slice_index)
```

```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
import os

def save_individual_regions(roi_image_path, output_dir):
    """
    Extracts each region from the ROI mask and saves it as a separate .nii.gz file.
    """
    # Load the ROI mask 3D image
    roi_img = nib.load(roi_image_path)
    roi_data = roi_img.get_fdata()

    # Find unique region identifiers, ignoring the background (0)
    regions = np.unique(roi_data)[1:]  # Skip 0 (background)
    
    # Ensure the output directory exists
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # Process and save each region
    for region in regions:
        # Create a mask for the current region
        region_mask = np.where(roi_data == region, region, 0)
        
        # Save the mask as a new NIfTI image
        region_img = nib.Nifti1Image(region_mask.astype(np.int16), roi_img.affine, roi_img.header)
        region_filename = os.path.join(output_dir, f'region_{int(region)}.nii.gz')
        nib.save(region_img, region_filename)
        print(f"Saved {region_filename}")

def visualize_regions(output_dir, slice_index):
    """
    Displays each region in its own subplot within a single figure.
    """
    # Get the list of saved region files
    region_files = [f for f in os.listdir(output_dir) if f.endswith('.nii.gz')]
    n_plots = len(region_files)
    
    cols = 13  # Number of columns in the subplot grid
    rows = n_plots // cols + (n_plots % cols > 2)  # Calculate rows needed
    
    # Create a figure for subplots
    fig, axes = plt.subplots(rows, cols, figsize=(15, rows * 1))
    if n_plots > 1:
        axes = axes.flatten()  # Flatten if more than one subplot
    else:
        axes = [axes]  # Wrap in list if only one subplot

    for i, region_file in enumerate(region_files):
        region_img = nib.load(os.path.join(output_dir, region_file))
        region_data = region_img.get_fdata()
        
        slice_index = region_img.shape[2] // 2
        
        # Plotting the 2D view of the current region
        axes[i].imshow(region_data[:, :, slice_index], cmap='gray', interpolation='nearest')
        axes[i].set_title(f'{region_file}')
        axes[i].axis('off')  # Hide axes for clarity
    
    # Hide any unused subplot areas
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')
    
    plt.tight_layout()
    plt.show()

# Example usage
roi_image_path = '/work/shared/ngmm/scripts/Taiabur/NG2564_SegmentsSeeds.nii'
output_dir = '/user1/ngmm/tr855969/Desktop/Taiabur/ngmm-nnunet/dataset/individual_regions/NG2564'
slice_index = 150  # Adjust based on the slice to visualize
# save_individual_regions(roi_image_path, output_dir)
visualize_regions(output_dir, slice_index)
```
