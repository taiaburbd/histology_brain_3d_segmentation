```python
import os
import torch
import numpy as np
from pathlib import Path
from torchvision.io import read_image
import glob
import SimpleITK as sitk

```

```python

# Path Setup
MAIN_PATH = Path('./').resolve()
DATA_PATH = MAIN_PATH / 'dataset'
```

```python

# Function to load dataset
def load_dataset(data_path):
    dataset = []
    subfolders = [folder for folder in data_path.iterdir() if folder.is_dir()]
    for subfolder in subfolders:
        rc_files = glob.glob(os.path.join(subfolder, "*_RC.nrrd"))
        brainmask_files = glob.glob(os.path.join(subfolder, "*_brainmask.seg.nrrd"))
        
        for rc_file, brainmask_file in zip(rc_files, brainmask_files):
            rc_image = sitk.ReadImage(rc_file)  # Read RC image
            brainmask_image = sitk.ReadImage(brainmask_file)  # Read brainmask image
            print(rc_image)
            print(brainmask_image)
            # dataset.append((rc_image, str(rc_file.relative_to(DATA_PATH))))  # Save RC image and its path
            # dataset.append((brainmask_image, str(brainmask_file.relative_to(DATA_PATH))))  # Save brainmask image and its path
            
    return dataset

# # Load dataset
dataset = load_dataset(DATA_PATH)
```
